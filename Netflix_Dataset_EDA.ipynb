**Netflix Dataset - Exploratory Data Analysis**

```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
```

```python
# 2. Upload the dataset
uploaded = files.upload()
```

```python
# 3. Load the CSV file
df = pd.read_csv("/content/Dataset.csv")
```

```python
# 4. Display the first 5 rows
print("First 5 rows:")
display(df.head())
```

```python
# 5. Check the dataset size
print("Dataset shape:", df.shape)
```

```python
# 6. Check column names and data types
print("\nDataset information:")
df.info()
```

```python
# 7. Check missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# 8. Check duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
# 9. Display statistical information
print("\nStatistical summary:")
display(df.describe(include="all"))
```

```python
# 10. Analyze Movies and TV Shows
if "type" in df.columns:
    print("\nContent type distribution:")
    print(df["type"].value_counts())

    df["type"].value_counts().plot(
        kind="bar",
        title="Number of Movies and TV Shows"
    )
    plt.xlabel("Content Type")
    plt.ylabel("Number of Titles")
    plt.show()
```

```python
# 11. Analyze release years
if "release_year" in df.columns:
    df["release_year"].value_counts().sort_index().plot(
        kind="line",
        figsize=(10, 5),
        title="Netflix Titles by Release Year"
    )
    plt.xlabel("Release Year")
    plt.ylabel("Number of Titles")
    plt.show()
```

```python
# 12. Analyze content ratings
if "rating" in df.columns:
    df["rating"].value_counts().plot(
        kind="bar",
        figsize=(10, 5),
        title="Distribution of Content Ratings"
    )
    plt.xlabel("Content Rating")
    plt.ylabel("Number of Titles")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# 13. Analyze the most common genres
if "listed_in" in df.columns:
    genres = df["listed_in"].dropna().str.split(", ").explode()

    print("\nTop 10 genres:")
    print(genres.value_counts().head(10))

    genres.value_counts().head(10).plot(
        kind="bar",
        figsize=(10, 5),
        title="Top 10 Netflix Genres"
    )
    plt.xlabel("Genre")
    plt.ylabel("Number of Titles")
    plt.xticks(rotation=45)
    plt.show()

print("\nDataset analysis completed successfully!")
```
